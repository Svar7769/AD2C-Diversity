# Diversity–reward landscape $J(\theta)$

Fixed-target DiCo for 1-, 2-, and 3-goal Navigation. Points are means across seeds and error bars are ± standard error.


```python
from diversity_reward_landscape import discover, download, aggregate, plot
```

```python
# Select the new 54-run series from benchmarl using W&B config values.
# Filters: model_config.desired_snd, task_config.agents_with_same_goal, and seed.
api, run_by_name, matched = discover()
```

```python
downloaded_files = download(run_by_name, matched)
print(f'{len(downloaded_files)} histories ready')
```

```python
# There are three seeds (0, 1, 2), so error bars use mean ± SEM over 3 seeds.
# Converged return = mean of the final 10% of evaluation points per seed.
per_run, landscape = aggregate(final_fraction=0.10)
display(landscape)
incomplete = landscape.loc[landscape.seeds != 3, ['goals', 'theta', 'seeds']]
if not incomplete.empty:
    print('WARNING: these points do not have all three independent seeds:')
    display(incomplete)
```

```python
fig, ax = plot(landscape)
```
